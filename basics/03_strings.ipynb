# 03 · Strings

**Learning goals**

- Write string literals with single, double and triple quotes.
- Use escape sequences.
- Read characters with indexes and slices.
- Understand why strings are immutable.

## Quotes

A string is text inside quotes. The trouble starts when the text itself
contains the quote character:

```python
message = 'Hello. I'm Mahdi'   # SyntaxError: unterminated string literal
```

The apostrophe in `I'm` ends the string early. Two fixes: use double quotes
around the text, or escape the apostrophe with a backslash.

````python
message = "Hello. I'm Mahdi"
message
````

````python
message == 'Hello. I\'m Mahdi'  # the escaped version is identical
````

## Indexing

Each character has a position, starting at **0**. Negative indexes count from
the end (`-1` is the last character). `len()` gives the length.

````python
message[4]
````

````python
message[-3]
````

````python
len(message)
````

Asking for a position that does not exist raises `IndexError`:

````python
try:
    message[20]
except IndexError as error:
    print(f"IndexError: {error}")
````

## Slicing

`text[start:stop:step]` returns a piece of the string. `start` is included,
`stop` is **excluded**, and any part may be left out. Unlike indexing, a slice
that runs past the end is silently clipped instead of raising an error.

````python
message[11:99]
````

````python
digits = "0123456789"
digits[3:7]
````

````python
digits[2:8:3]  # every 3rd character from index 2 up to index 7
````

````python
digits[1:-1]  # drop the first and last character
````

````python
digits[4:]  # from index 4 to the end
````

````python
digits[:-3]  # everything except the last three
````

````python
digits[::-1]  # step -1 walks backwards: the string reversed
````

## Multi-line strings and escape sequences

Triple quotes can span several lines and may contain both kinds of quotes.
Inside ordinary quotes, a backslash starts an *escape sequence*: `\n` is a
new line, `\"` a double quote and `\'` an apostrophe.

````python
text = """Hello,
I'm Mahdi.
Say "hi"!"""
text
````

````python
print("My name is Mahdi\nAnd this is a \"test\"")
````

## Strings are immutable

You can read a character but not assign to one. To "change" a string, build
a **new** one from pieces and give it the same name.

````python
name = "Mahdi"
name[0]
````

````python
try:
    name[0] = "m"
except TypeError as error:
    print(f"TypeError: {error}")
````

````python
name = "m" + name[1:]
name
````

## Repetition

`+` joins strings together and `*` repeats them.

````python
name * 3
````

````python
(name + " ") * 3
````

## Key takeaways

- Choose quotes (or escapes) so the text does not end the literal early.
- Indexes start at 0; negative indexes count from the end.
- Slices are `[start:stop:step]` with `stop` excluded, and never raise for
  out-of-range bounds.
- Strings cannot be modified in place; build a new string instead.
