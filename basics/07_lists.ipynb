## Creating and combining lists

A list is an ordered, changeable sequence. It can hold values of different
types. `+` joins two lists into a new one and `*` repeats a list.

```python
numbers = [0, 1, 2, 3]
animals = ['bird', 'cat', 'lion', 'dog', 3.14, False, 5]
print(numbers + animals)
```

```python
repeated = numbers * 2
repeated
```

## Reading items

Indexing works like strings: it starts at 0, and a missing index raises
`IndexError`.

```python
repeated[4]
```

```python
try:
    repeated[10]
except IndexError as error:
    print(f"IndexError: {error}")
```

## Counting and length

`count(x)` returns how many times `x` appears, and `len()` returns the number
of items.

```python
repeated.count(2)
```

```python
repeated.count(20)
```

```python
len(repeated)
```

## Removing the last item with `pop()`

`pop()` **removes and returns** the last item, so the list gets shorter and
you can keep the value.

```python
repeated.pop()
```

```python
repeated
```

```python
last = animals.pop()
last
```

```python
animals
```

## Changing items in place

Lists are mutable: you can assign to a position, remove a value, and sort or
reverse the whole list.

- `sort()` orders the items **in place** and returns `None`.
- `remove(x)` deletes the first item equal to `x`.
- `reverse()` flips the order in place.

```python
animals = ['lion', 'cat', 'bird', 'dog']
animals.sort()
animals
```

```python
animals[2] = "fish"
animals
```

```python
animals.remove('cat')
animals
```

```python
animals.reverse()
animals
```

A common mistake is to assign the result of an in-place method. Because
`sort()` returns `None`, this destroys the list:

```python
mistake = ['b', 'a'].sort()
print(mistake)
```

## Key takeaways

- Lists are ordered and mutable; `+` and `*` create new lists.
- `pop()` returns the removed item; `sort()`, `reverse()` and `remove()` change
  the list in place.
- In-place methods return `None`, so do not assign their result.
