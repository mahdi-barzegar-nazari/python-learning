# 09 · Dictionaries

**Learning goals**

- Store and look up values by key.
- Nest dictionaries and lists.
- Use `keys()`, `values()`, `items()` and `get()`.

**Practice:** [10 · Dictionary practice](10_dict_practice.py)

## Key → value

A dictionary maps **keys** to **values**. You look a value up with its key
instead of a numeric position.

```python
stock = {"orange": 24, "kiwi": 15, "lemon": 8}
stock["kiwi"]
```

## Nested dictionaries

Values can be anything, including other dictionaries. Chain the keys to reach
inside.

```python
baskets = {
    "amir": {"orange": 4, "kiwi": 16, "lemon": 9},
    "reza": {"kiwi": 11, "orange": 8, "lemon": 3},
}
```

```python
baskets["amir"]["lemon"]
```

## Keys, values and items

These three methods return *views* of the dictionary: `keys()` for the keys,
`values()` for the values and `items()` for `(key, value)` pairs.

```python
print(baskets.keys())
```

```python
print(baskets.values())
```

```python
print(baskets.items())
```

```python
price = {"pear": 60, "mango": 100, "grape": 80}
```

```python
price.keys()
```

```python
price.values()
```

```python
price.items()
```

## Safe lookups with `get()`

Indexing with a missing key raises `KeyError`. `get(key, default)` returns
the default instead (`None` if you give none), which avoids the crash.

```python
try:
    price["peach"]
except KeyError as error:
    print(f"KeyError: {error}")
```

```python
price.get("mango")
```

```python
price.get("peach", -1)
```

```python
price.get("grape", -1)
```

## Lists inside dictionaries

Combine the two by chaining a key and then an index.

```python
data = {"list1": [1, 2, 3, 4], "list2": [7, 13, 3.14]}
data["list2"]
```

```python
data["list2"][1]
```

```python
data["list2"][1] + 4
```

## Key takeaways

- `d[key]` reads a value and raises `KeyError` if the key is missing;
  `d.get(key, default)` does not.
- `keys()`, `values()` and `items()` let you loop over a dictionary.
- Dictionaries and lists nest freely: chain `[key]` and `[index]`.
