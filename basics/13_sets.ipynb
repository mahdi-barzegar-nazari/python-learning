# 13 · Sets

**Learning goals**

- Create a set and add items to it.
- Explain why a set has no duplicates and no order.
- Combine sets with `union()` and `intersection()`.
- Recognise the "I named my variable `set`" bug and know how to avoid it.

## An unordered collection of unique items

A set is written with `{}` or built with `set()`. Unlike a list, it cannot
contain the same value twice, and it does not remember insertion order.
`add()` puts one item in.

```python
myset = set()
myset.add(2)
myset.add(5)
myset
```

```python
type(myset)
```

## Duplicates are silently ignored

Adding a value that is already there does nothing; the set does not grow and
raises no error.

```python
myset.add(2)
myset  # still {2, 5}
```

## `add()` takes exactly one item

A common mistake is calling `add()` the way you might call `print()`, with
several values. It only accepts **one** argument, so `add(1, 3)` raises
`TypeError`.

```python
try:
    myset.add(1, 3)
except TypeError as error:
    print(f"TypeError: {error}")
```

Passing a single list does not work either, for a different reason: every
set item must be *hashable* (roughly: unchangeable), and a list is mutable.

```python
try:
    myset.add([1, 3])
except TypeError as error:
    print(f"TypeError: {error}")
```

To add several items at once, use `update()` instead, which accepts any
sequence.

```python
myset.update([1, 3])
myset
```

## Combining sets

`union()` returns every item that is in **either** set. `intersection()`
returns only the items that are in **both**. Neither method changes the
original sets.

```python
math = set()
math.update(["reza", "amir"])
math
```

```python
chess = set()
chess.update(["amir", "hesam"])
chess
```

```python
math.union(chess)
```

```python
math.intersection(chess)
```

## Building a set from a list

Passing a list to `set()` is a quick way to drop duplicates, because the
result can only keep one copy of each value.

```python
fruit = set(["apple", "banana", "kiwi", "lemon", "apple"])
fruit  # the repeated 'apple' collapses to one
```

```python
fruit.add("orange")
fruit
```

> **The printed order is not something to rely on.** A set has no order, so
> Python is free to display its items in any sequence, and for a set of
> strings that sequence can even change between runs of your program. If you
> need a fixed order, convert with `sorted(fruit)`.

## Danger: do not name a variable `set`

`set` is also the name of the **built-in type**. Assigning to it replaces
that type for the rest of the program, which breaks every later call to
`set()` in a very confusing way.

```python
set = {1, 2}  # DANGER: this shadows the built-in set() constructor
set
```

```python
try:
    set()
except TypeError as error:
    print(f"TypeError: {error}")
```

`del` removes our variable, which uncovers the built-in again.

```python
del set
set()  # noqa: F821 # the real constructor works again
```

## Key takeaways

- `{}`/`set()` store unique, unordered items; duplicates are dropped
  silently.
- `add()` takes exactly one hashable item; `update()` takes many.
- `union()` and `intersection()` combine two sets without changing either.
- Never name a variable (or loop variable) `set`, `list`, `dict`, `str` and
  so on. It shadows the built-in type and later code that expects the real
  one will break with a confusing `TypeError`.
