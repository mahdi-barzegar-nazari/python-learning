## An ordered, unchangeable sequence

A tuple looks like a list but uses parentheses and **cannot be modified**
after it is created. Reading works exactly like a list.

```python
t1 = (3, 2, 1, "test", 32.124)
t1
```

```python
t1[2]
```

```python
len(t1)
```

## Tuples are immutable

Assigning to an item raises `TypeError`. Use a tuple for data that should
stay fixed, such as coordinates or a `(name, score)` record.

```python
try:
    t1[2] = 5
except TypeError as error:
    print(f"TypeError: {error}")
```

## Unpacking

If you have as many variables as items, you can assign them all in one line.
This is often clearer than indexing.

```python
result = ("Ali", 18)
name, score = result
```

```python
score
```

```python
name
```

## Key takeaways

- Tuples are like lists that cannot change.
- `name, score = result` unpacks a tuple into variables.
- Unpacking is clearer than `result[0]` and `result[1]`.
