## Integers and basic arithmetic

Python evaluates arithmetic the way you would on paper: `+` adds, `-`
subtracts and `*` multiplies. Negative numbers work as you would expect.

```python
-456 + 654
```

```python
8 * 14
```

`print()` writes text to the output area. Unlike a bare expression, it shows
the value without quotes and can print several things in a row.

```python
print("Hiii")
```

## Division: three different operators

| Operator | Name | Example | Result |
| :--- | :--- | :--- | :--- |
| `/` | true division (always a `float`) | `15 / 4` | `3.75` |
| `//` | floor division (drops the fraction) | `15 // 4` | `3` |
| `%` | remainder (modulo) | `15 % 4` | `3` |

Notice that `/` returns a `float` even when the division is exact.

```python
20 / 5
```

```python
15 / 4
```

```python
15 // 4
```

```python
15 % 4
```

`//` rounds **down** (toward negative infinity), not toward zero. This only
shows up with negative numbers:

```python
-7 // 2
```

## Remainder tricks

A number is divisible by `k` exactly when `n % k == 0`. Checking `n % 2 == 0`
is the standard test for an even number.

```python
957 % 3
```

```python
print(957 % 3 == 0)  # divisible by 3?
print(376 % 2 == 0)  # even?
```

## Powers and very large numbers

`**` is the power operator. Python integers have no fixed size limit, so
results that would overflow in many other languages are still exact.

```python
7**4
```

```python
34**45
```

## Floats are approximations

Numbers with a decimal point are stored as binary floating-point values, so
many decimal fractions cannot be represented exactly.

```python
87.26 / 42.8
```

```python
0.1 + 0.2
```

## Readable numbers and operator precedence

Underscores in a number literal are ignored, which makes long numbers easier
to read. Multiplication and division are evaluated before addition and
subtraction; use parentheses to make the order explicit.

```python
9_999_999
```

```python
2 + 3 * 4
```

```python
(2 + 3) * 4
```

```python
((4 + 3) * (8 / 2)) * 5
```

The last result is `140.0`, not `140`, because `8 / 2` produced a float.

## Errors are information: division by zero

Some operations are impossible. Python stops and raises an *exception*. Here
we **catch** it with `try` / `except` so the notebook keeps running and we can
read the message.

```python
try:
    7 / 0
except ZeroDivisionError as error:
    print(f"ZeroDivisionError: {error}")
```

## Key takeaways

- `/` gives a float, `//` gives the floored quotient, `%` gives the remainder.
- Use parentheses whenever the order of operations is not obvious.
- Integers are exact at any size; floats are approximations.
- Dividing by zero raises `ZeroDivisionError`.
