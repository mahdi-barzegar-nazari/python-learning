**Learning goals**

- Recognise the basic types: `str`, `int` and `float`.
- See how the same operator behaves differently on different types.
- Use common string methods and remember they return **new** strings.

## Types

`type()` tells you what kind of value you have. The type decides what
operations are allowed and what they mean.

```python
name = "Mahdi"
age = 99
height = 198.7
```

```python
type(name)
```

```python
type(age)
```

```python
type(height)
```

The same operator can mean different things. With two strings `+` joins
them; with two numbers it adds them.

```python
"4" + "6"
```

```python
4 + 6
```

## Changing case

`lower()` and `upper()` return a converted copy. `isupper()` answers a yes or
no question with `True` or `False`.

```python
name.lower()
```

```python
name.upper()
```

```python
name.isupper()
```

## Methods return new strings

Strings are immutable (see notebook 03), so `name.upper()` does **not** change
`name`. Assign the result back if you want to keep it.

```python
name
```

```python
name = name.upper()
name
```

## Searching: `index` and `find`

`index()` returns the position of a substring. The search is **case
sensitive**, and if the text is missing it raises `ValueError`. `find()` does
the same job but returns `-1` instead of raising an error.

```python
"Mahdi".index("h")
```

```python
try:
    "Mahdi".index("H")
except ValueError as error:
    print(f"ValueError: {error}")
```

```python
"Mahdi".find("H")
```

## Splitting

`split(separator)` cuts a string into a **list** of pieces. With no argument
it splits on any run of whitespace.

```python
animals = "dog,cat,bird,fish,cow,lion"
animals.split(",")
```

## Key takeaways

- `type()` reveals a value's type; operators depend on the type.
- String methods return new strings; the original is unchanged.
- `index()` raises `ValueError` when nothing is found, `find()` returns `-1`.
- `split()` turns a string into a list.
