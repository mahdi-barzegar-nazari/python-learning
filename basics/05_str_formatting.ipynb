# 05 · String formatting

**Learning goals**

- Build strings that contain variables in four different ways.
- Control the number of decimals.
- Know which style to prefer today (f-strings).

## Setup

We will build the same greeting several ways.

```python
name = "Mahdi"
family_name = "Barzegar"
age = 99.78215403
```

## 1. Concatenation

Join pieces with `+`. It works, but every space has to be typed by hand and
numbers must be converted with `str()` first.

```python
"Hello " + name + " " + family_name
```

## 2. The `%` operator (old style)

Placeholders such as `%s` (string) and `%i` (integer) are filled from a
tuple. Notice that `%i` **truncates** the float: `99.78` becomes `99`.

```python
"Hello %s %s, you are %i" % (name, family_name, age)
```

## 3. `str.format()`

Placeholders are `{}`. Numbers select an argument by position, so the order
can differ from the order in the text. Names make the template self-explaining,
and a *format spec* after a colon controls the output: `.1f` means one digit
after the decimal point and **rounds** (`99.78` becomes `99.8`).

```python
"Hello {1}, {0} you are {2}".format(name, family_name, age)
```

```python
print("Hello {first} {last}, you are {years:.1f}".format(
    first=name, last=family_name, years=age
))
```

## 4. f-strings (recommended)

Put an `f` in front of the quotes and write expressions directly inside
`{}`. They are the shortest and easiest to read, and they accept the same
format specs.

```python
title = "Mr. "
print(f"Hello {title + name} {family_name}, "
      f"you are {age:.1f} years old")
```

## Key takeaways

| Style | Best for |
| :--- | :--- |
| `+` | very short strings |
| `%` | reading older code |
| `.format()` | templates stored separately from the values |
| f-string | almost everything else |

- `%i` truncates a float; `:.1f` rounds it.
